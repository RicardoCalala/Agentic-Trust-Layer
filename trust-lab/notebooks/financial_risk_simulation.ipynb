# Financial-risk simulation prototype

This notebook models only fictional, aggregate-first signals. It is a UI prototyping aid—not a detection system and not a basis for action against any person.

```python
profiles = [
    {'label': 'Aggregate variance', 'score': 31, 'confidence': 56, 'sources': 2},
    {'label': 'Cross-source mismatch', 'score': 48, 'confidence': 63, 'sources': 3},
    {'label': 'Coverage change', 'score': 24, 'confidence': 51, 'sources': 2},
]
profiles
```

```python
def next_frame(frame):
    profile = profiles[frame % len(profiles)]
    return {**profile, 'stage': frame % 6 + 1, 'decision': 'human review remains required'}

[next_frame(frame) for frame in range(6)]
```
